#### 收入

##### 公司在一定时间内销售商品/服务获得的收入或毛收入

#### 销售成本

##### 和公司产品/服务销售之间直接关联的成本，如生产这个商品的材料费用

#### 毛利润

##### 1. 这是一个合计类目
##### 2. 计算方式：sum(收入 - 销售成本)
#### 如：一家汽车厂商：

##### 1. 一辆车售价为 $20,000

##### 2. 生产该车的成本为 $5,000
##### 那么他的毛利润，如下

```python
price_per_car = 20000
cost_per_car = 5000

gross_profit_per_car = price_per_car - cost_per_car

print(f'汽车毛利润：${gross_profit_per_car}')
```

```python
qtr_cars_sold_count = 25  # 这个季度中销售的汽车数

income_qtr = qtr_cars_sold_count * price_per_car
cost_qtr = cost_per_car * qtr_cars_sold_count
gross_profit_qtr = qtr_cars_sold_count * gross_profit_per_car

print(f'这个季度收入：${income_qtr}')
print(f'这个季度的销售成本：${cost_qtr}')
print(f'这个季度的毛利润：${gross_profit_qtr}')
```

```python
gross_margin = gross_profit_per_car / price_per_car * 100
print(f'毛利率：{gross_margin}%')
```

#### 营业费用

##### 支撑公司正常运转产生的费用，并且可以细分多个不同的分类：

##### * 销售、管理、行政：Sales, Govern, Adaministration

##### * 营销：Marketing

##### * 研发：Research and Development

#### 这家汽车公司，还有以下营业费用方面的开支：

```python
expenses = {
    'sales_and_govern': 77500,
    'markerting': 7500,
    'research_and_development': 5000,
}

operating_expense = sum([expenses[item] for item in expenses.keys()])

print(f'营业费用：${operating_expense}')
```

#### 其他收益

##### 1. 并不是公司核心业务产生的收入，如：这家汽车公司除了销售汽车，还进行金融融资，每年获取粉分红
##### 2. 这种收入不是核心业务产生的，所以不计入收入这一项
##### 3. 对于公司来说，这种收入仍然需要纳税

#### EBITDA

##### 1. 息、税、折旧、摊销前的利润
##### 2. 华尔街分析师关注的重要指标，广泛用于分析和估值
